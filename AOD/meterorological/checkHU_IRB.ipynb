```python
import pandas as pd

df = pd.read_csv("../../data/HU_IRB_AOD500_preprocessed.csv", parse_dates=["datetime"])

print("Null values:\n", df.isnull().sum())
print("\nShape:", df.shape)
display(df.head())

# 1-hour average
df_hourly = df.set_index("datetime").resample("1h").mean().reset_index()

print("-----------"*80)

print("Null values:\n", df.isnull().sum())
print("Shape:", df_hourly.shape)
display(df_hourly.head())
```
